```python
from ete4 import PhyloTree
from ete4.smartview import Layout, BASIC_LAYOUT


t = PhyloTree(open("PF13853_9606.7764.mafft.lg.treefile.rooted.withinternalnames"), parser=1)

outgroup_leaf_name = 'New|7764.A0A8C4N3S4'
t.set_outgroup(outgroup_leaf_name)

# 7764 outgroup 
outgroup_styles = {}
blue = {"stroke-width": 2.3, "stroke": "#a8c435"}

for n in t.traverse():
    if n.is_leaf:
        header = n.name.split("|", 1)[1] if n.name.startswith("New|") else n.name
        if header.split(".", 1)[0] == "7764":
            cur = n
            while cur:
                outgroup_styles[cur] = blue
                cur = cur.up
            break

def draw_7764(node):
    style = outgroup_styles.get(node)
    return {"hz-line": style, "vt-line": style} if style else {}

layout_outgroup = Layout("7764 Branch Highlight", draw_node=draw_7764)

# two child sub-trees of node_1  
node1_styles = {}

node1 = [n for n in t.traverse() if n.name == "node_1"][0] 
child_a, child_b = node1.children                           

style_a = {"stroke-width": 2.3, "stroke": "#7f1734"}
style_b = {"stroke-width": 2.3, "stroke": "#557c99"}

for node in child_a.traverse():
    for child in node.children:       
        node1_styles[child] = style_a
        
for node in child_b.traverse():
    for child in node.children:       
        node1_styles[child] = style_b

def draw_node1_children(node):
    style = node1_styles.get(node, None)
    return {"hz-line": style, "vt-line": style} if style else {}

layout_node1 = Layout("Class I,II Colouring", draw_node=draw_node1_children)

# hexagons to desired nodes 
hexagon_style = {"dot": {"shape": "circle", "radius": 8, "fill": "#1a1919", "stroke": "black"},
                 "vt-line": {"stroke": "#1a1919", "stroke-width": 5, "stroke-dasharray": "5,5"}}

def draw_hexagons(node):
    return hexagon_style if node.name in {"node_1", "node_2", "node_3"} else {}

layout_hex = Layout("Key Nodes Hexagon", draw_node=draw_hexagons)

# explore
t.explore(layouts=[layout_outgroup, layout_node1, layout_hex, BASIC_LAYOUT])
```
