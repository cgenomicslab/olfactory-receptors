```python
from pathlib import Path
import pandas as pd
```

```python
fasta_path = Path("../data/ReferenceTree/PF13853.9606_7955_7740_7764_75743_137246.fa")
```

```python
human_seqs = {} 

with open(fasta_path, "r") as f:
    header = None
    seq_lines = []
    for line in f:
        line = line.rstrip("\n")
        if line.startswith(">"):
            if header and header.startswith(">9606."):
                human_seqs[header] = "".join(seq_lines)
            header = line
            seq_lines = []
        else:
            seq_lines.append(line)
    if header and header.startswith(">9606."):
        human_seqs[header] = "".join(seq_lines)
        
for i, k in enumerate(list(human_seqs.keys())[:5], 1):
    print(f"{k} : {(human_seqs[k])}")
```

```python
#write those 433 fasta sequences to a new fasta file
with open("human_ORs_tree/human_433_seqs.fa", "w") as f:
    for header, seq in human_seqs.items():
        f.write(f"{header}\n{seq}\n")   
```
