### create unique identifiers for proteins & ligands

```python
seq_codes, seq_uniques = pd.factorize(m2or['Sequence'])
smiles_codes, smiles_uniques = pd.factorize(m2or['SMILES'])

m2or['seq_id'] = [f'SEQ{code:04d}' for code in seq_codes]
m2or['smiles_id'] = [f'SML{code:04d}' for code in smiles_codes]

seq_mapping = pd.DataFrame({'seq_id': [f'SEQ{idx:04d}' for idx in range(len(seq_uniques))],
                           'Sequence': seq_uniques})

smiles_mapping = pd.DataFrame({'smiles_id': [f'SML{idx:04d}' for idx in range(len(smiles_uniques))],
                              'SMILES': smiles_uniques})
```

### create boolean column for Mutation

```python
# add a new boolean column "mutation", having True if "Mutation" is not null
pairs['mutation'] = pairs['Mutation'].notnull()
```

```python
pairs.drop(columns=["Mutation"], inplace=True)
```

### create class 1/class 2 based on phylogeny

```python
unique_sequences = pairs[['seq_id', 'Sequence']].drop_duplicates().set_index('seq_id')
with open('../processed/m2or_sequences_phylogeny.fasta', 'w') as f:
    for seq_id, sequence in unique_sequences.itertuples(index=True):
        f.write(f'>{seq_id}\n{sequence}\n')
```

```python
from ete4 import PhyloTree

t = PhyloTree("../processed/m2or_sequences_phylogeny_msa_gt01_fasttree.nw")
midp = t.get_midpoint_outgroup()
t.set_outgroup(midp)

children = list(t.children)
child1, child2 = children[0], children[1]

headers_child1 = [leaf.name for leaf in child1.leaves()]
headers_child2 = [leaf.name for leaf in child2.leaves()]

with open("../processed/child1_headers.txt", "w") as f:
    for h in headers_child1:
        f.write(h + "\n")
with open("../processed/child2_headers.txt", "w") as f:
    for h in headers_child2:
        f.write(h + "\n")

t.explore()
```

```python
with open("../processed/child1_headers.txt") as f:
    child1_set = set(line.strip() for line in f if line.strip())
with open("../processed/child2_headers.txt") as f:
    child2_set = set(line.strip() for line in f if line.strip())

def get_class(seq_id):
    if seq_id in child1_set:
        return "class 1"
    elif seq_id in child2_set:
        return "class 2"
    else:
        return "other"  

pairs["Class"] = pairs["seq_id"].apply(get_class)
```

```python
print("unique sequencess:", pairs["Sequence"].nunique())
print("unique SMILES:", pairs["SMILES"].nunique())
print("Counts:", pairs["Responsive"].value_counts())
print("Mutated sequences:", pairs[pairs["mutation"] == True]["Sequence"].nunique())
print("Wild-type sequences:", pairs[pairs["mutation"] == False]["Sequence"].nunique())
print("Class 1 sequences:", pairs[pairs["Class"] == "class 1"]["Sequence"].nunique())
print("Class 2 sequences:", pairs[pairs["Class"] == "class 2"]["Sequence"].nunique())

inactive_sequences = pairs.groupby("Sequence")["Responsive"].sum().loc[lambda s: s == 0].index.tolist()
print("inactive sequences:", len(inactive_sequences))
```
